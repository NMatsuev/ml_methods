**Задание 3**. В списке, состоящем из целых чисел, вычислить:
*   количество элементов массива, больших С;
*   произведение элементов массива, расположенных после максимального по модулю элемента;
*   удалить все положительные элементы списка.

```python
numbers = list(map(int, input("Введите элементы списка через пробел: ").split()))
C = int(input("Введите C: "))

# Количество элементов, больших C
count = 0

for number in numbers:
    if number > C:
        count += 1

print("Количество элементов, больших C:", count)

# Поиск максимального по модулю элемента
max_index = 0

for i in range(1, len(numbers)):
    if abs(numbers[i]) > abs(numbers[max_index]):
        max_index = i

# Произведение элементов после максимального по модулю
if max_index == len(numbers) - 1:
    product = 0
else:
    product = 1

    for i in range(max_index + 1, len(numbers)):
        product *= numbers[i]

print("Произведение элементов после максимального по модулю:", product)

# Удаление положительных элементов
numbers = [number for number in numbers if number <= 0]

print("Список после удаления положительных элементов:", numbers)
```

**Задание 5**. Реализуйте программу «Ювелирный магазин», которая будет
включать в себя шесть пунктов меню.  У вас есть словарь, где ключ –
название изделия. Значение – список, который содержит состав
изделия(золото, серебро,и т.п.), цену и кол-во (шт),которое есть в
магазине.  
1. Просмотр описания: название – описание
2. Просмотр цены: название – цена.
3. Просмотр количества: название – количество.
4. Всю информацию.
5. Покупка
В пункте «Покупка» необходимо совершить покупку,  
с клавиатуры вводите название изделия и его кол-во, n – выход из
программы. Посчитать цену выбранных товаров и сколько товаров
осталось в изначальном списке.       
6. До свидания  

```python
products = {
    "Кольцо": ["Золото 585 пробы", 25000, 5],
    "Серьги": ["Серебро 925 пробы", 12000, 8],
    "Цепочка": ["Золото 585 пробы", 35000, 4],
    "Браслет": ["Серебро 925 пробы", 15000, 6],
    "Кулон": ["Золото 750 пробы", 30000, 3]
}

while True:
    print("\n--- ЮВЕЛИРНЫЙ МАГАЗИН ---")
    print("1. Просмотр описания")
    print("2. Просмотр цены")
    print("3. Просмотр количества")
    print("4. Всю информацию")
    print("5. Покупка")
    print("6. До свидания")

    choice = input("Выберите пункт меню: ")

    # 1. Описание
    if choice == "1":
        name = input("Введите название изделия: ")

        if name in products:
            print("Описание:", products[name][0])
        else:
            print("Такого изделия нет.")

    # 2. Цена
    elif choice == "2":
        name = input("Введите название изделия: ")

        if name in products:
            print("Цена:", products[name][1], "руб.")
        else:
            print("Такого изделия нет.")

    # 3. Количество
    elif choice == "3":
        name = input("Введите название изделия: ")

        if name in products:
            print("Количество:", products[name][2], "шт.")
        else:
            print("Такого изделия нет.")

    # 4. Вся информация
    elif choice == "4":
        for name, info in products.items():
            print("\nНазвание:", name)
            print("Описание:", info[0])
            print("Цена:", info[1], "руб.")
            print("Количество:", info[2], "шт.")

    # 5. Покупка
    elif choice == "5":
        while True:
            name = input("\nВведите название изделия (n - выход): ")

            if name.lower() == "n":
                break

            if name not in products:
                print("Такого изделия нет.")
                continue

            quantity = int(input("Введите количество: "))

            if quantity <= 0:
                print("Количество должно быть больше нуля.")
            elif quantity > products[name][2]:
                print("Недостаточно изделий на складе.")
            else:
                price = products[name][1]
                total = price * quantity

                products[name][2] -= quantity

                print("Покупка совершена.")
                print("Стоимость покупки:", total, "руб.")
                print("Осталось на складе:", products[name][2], "шт.")

    # 6. Выход
    elif choice == "6":
        print("До свидания!")
        break

    else:
        print("Неверный пункт меню.")
```

**Задание 1**. Написать функцию, которая  считает сумму четных цифр введенного
натурального числа.

```python
def sum_even_digits(number):
    total = 0

    while number > 0:
        digit = number % 10

        if digit % 2 == 0:
            total += digit

        number //= 10

    return total


number = int(input("Введите натуральное число: "))

print("Сумма четных цифр:", sum_even_digits(number))
```

**Задание 3**. Напишите функцию, которая будет принимать один аргумент. Если
множество, вывести первые три наибольших значения. Число – определить простое, или нет. Строка – посчитать количество гласных. Удалить все знаки препинания
(, . ! ?). Сделать проверку со всеми этими случаями.

```python
def process_data(data):

    # Если передано множество
    if isinstance(data, set):
        if len(data) == 0:
            print("Множество пустое.")
        else:
            largest = sorted(data, reverse=True)[:3]
            print("Три наибольших значения:", largest)

    # Если передано число
    elif isinstance(data, int):
        if data < 2:
            print("Число не является простым.")
        else:
            is_prime = True

            for i in range(2, int(data ** 0.5) + 1):
                if data % i == 0:
                    is_prime = False
                    break

            if is_prime:
                print("Число простое.")
            else:
                print("Число не является простым.")

    # Если передана строка
    elif isinstance(data, str):
        vowels = "аеёиоуыэюяАЕЁИОУЫЭЮЯ"

        vowel_count = 0

        for char in data:
            if char in vowels:
                vowel_count += 1

        for char in ",.!?":
            data = data.replace(char, "")

        print("Количество гласных:", vowel_count)
        print("Строка без знаков препинания:", data)

    else:
        print("Неизвестный тип данных.")


# Проверка со множеством
my_set = {10, 5, 25, 8, 17, 30}
process_data(my_set)

print()

# Проверка с числом
number = int(input("Введите число для проверки на простоту: "))
process_data(number)

print()

# Проверка со строкой
text = input("Введите строку: ")
process_data(text)
```

**Задание 1**. Создать класс Circle (круг). Поле класса хранит радиус окружности. Методы класса возвращают площадь и длину окружности. Выполнить проверку на то, что радиус-величина положительная.

```python
import math

class Circle:
    def __init__(self, radius):
        if radius <= 0:
            raise ValueError("Радиус должен быть положительным.")

        self.radius = radius

    def area(self):
        """Возвращает площадь круга."""
        return math.pi * self.radius ** 2

    def circumference(self):
        """Возвращает длину окружности."""
        return 2 * math.pi * self.radius


try:
    radius = float(input("Введите радиус: "))

    circle = Circle(radius)

    print("Площадь круга:", circle.area())
    print("Длина окружности:", circle.circumference())

except ValueError as error:
    print("Ошибка:", error)
```

**Задание 2**. Создать класс Employee (сотрудник) с полями ФИО, стаж, часовая
заработная плата, количество отработанных часов, оклад, премия. Создать
список сотрудников компании. Реализовать ввод данных всех сотрудников с
клавиатуры. Рассчитать с помощью методов класса заработную плату за
отработанное время, и премию, размер которой определяется в зависимости
от стажа работника (при стаже до 1 года 1%, до 3 лет 5%, до 5 лет 8%, свыше
5 лет 15%). С помощью метода печати, реализовать вывод информации о
работнике на экран.

```python
class Employee:
    def __init__(self, name, experience, hourly_rate, hours, salary):
        self.name = name
        self.experience = experience
        self.hourly_rate = hourly_rate
        self.hours = hours
        self.salary = salary
        self.bonus = 0

    def calculate_payment(self):
        """Расчет заработной платы за отработанное время."""
        return self.hourly_rate * self.hours

    def calculate_bonus(self):
        """Расчет премии в зависимости от стажа."""
        payment = self.calculate_payment()

        if self.experience < 1:
            percent = 0.01
        elif self.experience < 3:
            percent = 0.05
        elif self.experience < 5:
            percent = 0.08
        else:
            percent = 0.15

        self.bonus = payment * percent

        return self.bonus

    def total_salary(self):
        """Общая заработная плата."""
        return self.calculate_payment() + self.salary + self.calculate_bonus()

    def print_info(self):
        """Вывод информации о сотруднике."""
        print("\n--- Информация о сотруднике ---")
        print("ФИО:", self.name)
        print("Стаж:", self.experience, "лет")
        print("Часовая ставка:", self.hourly_rate)
        print("Отработано часов:", self.hours)
        print("Оклад:", self.salary)
        print("Заработок за отработанное время:", self.calculate_payment())
        print("Премия:", self.calculate_bonus())
        print("Итоговая заработная плата:", self.total_salary())


employees = []

count = int(input("Введите количество сотрудников: "))

for i in range(count):
    print("\nСотрудник №", i + 1)

    name = input("Введите ФИО: ")
    experience = float(input("Введите стаж (лет): "))
    hourly_rate = float(input("Введите часовую заработную плату: "))
    hours = float(input("Введите количество отработанных часов: "))
    salary = float(input("Введите оклад: "))

    employee = Employee(
        name,
        experience,
        hourly_rate,
        hours,
        salary
    )

    employees.append(employee)


print("\n========== СПИСОК СОТРУДНИКОВ ==========")

for employee in employees:
    employee.calculate_bonus()
    employee.print_info()
```

**Задание 3**. Реализовать класс Stationery (канцелярская принадлежность).
определить в нём атрибут title (название) и метод draw (отрисовка). Метод
выводит сообщение «Запуск отрисовки»; создать три дочерних класса Pen
(ручка), Pencil (карандаш), Handle (маркер); в каждом классе реализовать
переопределение метода draw. Для каждого класса метод должен выводить
уникальное сообщение; создать экземпляры классов и проверить, что
выведет описанный метод для каждого экземпляра.

```python
class Stationery:
    def __init__(self, title):
        self.title = title

    def draw(self):
        print("Запуск отрисовки")


class Pen(Stationery):
    def draw(self):
        print("Ручка: рисуем синими чернилами.")


class Pencil(Stationery):
    def draw(self):
        print("Карандаш: рисуем графитовым стержнем.")


class Handle(Stationery):
    def draw(self):
        print("Маркер: рисуем яркой линией.")


# Создаем экземпляры классов
stationery = Stationery("Канцелярская принадлежность")
pen = Pen("Ручка")
pencil = Pencil("Карандаш")
handle = Handle("Маркер")


# Проверяем работу метода draw()
print(stationery.title)
stationery.draw()

print("\n" + pen.title)
pen.draw()

print("\n" + pencil.title)
pencil.draw()

print("\n" + handle.title)
handle.draw()
```

**Задание 4**. Придумать класс самостоятельно, реализовать в нем методы экземпляра класса, статические, методы, методы класса.

```python
class Student:
    university = "Мой университет"

    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course

    # Метод экземпляра
    def show_info(self):
        print("Имя:", self.name)
        print("Возраст:", self.age)
        print("Курс:", self.course)
        print("Университет:", self.university)

    # Метод класса
    @classmethod
    def from_string(cls, data):
        name, age, course = data.split(";")

        return cls(name, int(age), int(course))

    # Статический метод
    @staticmethod
    def is_adult(age):
        return age >= 18


# Создаем объект обычным способом
student1 = Student("Иван", 19, 2)

print("Информация о первом студенте:")
student1.show_info()

print("\nПроверка возраста:")
print(Student.is_adult(student1.age))


# Создаем объект с помощью метода класса
student2 = Student.from_string("Анна;20;3")

print("\nИнформация о втором студенте:")
student2.show_info()

print("\nПроверка возраста:")
print(Student.is_adult(student2.age))
```
